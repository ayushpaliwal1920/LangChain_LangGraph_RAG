```python
from abc import ABC , abstractmethod
```

```python
class Runnable(ABC) : # abstract class

    @abstractmethod
    def invoke(input_data):
        pass
    
```

```python
import random

# Dummy LLm class / component: not logical just a demo

class NakliLLM(Runnable): # inherits from runnable(abstract class)

    def __init__(self):
        print("LLM created")

    # because we are inherting abstract(Runnable) class so we have to implement its abstracts methods also...

    def invoke(self , prompt):
            response_list = [
                "Delhi is the capital of INDIA.",
                "Ipl is a cricket league.",
                "Off campus is not that good except for lead female cast."
            ]
            
            return {
            "response" : random.choice(response_list)
            }
       

    def predict(self, prompt):

        response_list = [
            "Delhi is the capital of INDIA.",
            "Ipl is a cricket league.",
            "Off campus is not that good except for lead female cast."
        ]

        return {
             "response" : random.choice(response_list)
             }
```

```python
llm = NakliLLM()
```

```python
# Dummy Prompt template class / Component : just for demo

class NakliPromptTemplate(Runnable) :  # inherits from runnable(abstract class)

    def __init__(self , template , input_variables):
        self.template = template
        self.input_variables = input_variables

    # because we are inherting abstract(Runnable) class so we have to implement its abstracts methods also...

    def invoke(self , input_dict):
        return self.template.format(**input_dict)
        
    def format(self , input_dict):
        return self.template.format(**input_dict)
```

```python
# Dummy StringGenerator Parser / Component : 

class NakliStrOutputParser(Runnable) :  # inherits from runnable(abstract class)

    def __init__(self):
        pass

    def invoke(self, input_data):
        return input_data["response"]
```

```python
# Connect two or more component class :

class RunnableConnector(Runnable) :

    def __init__(self , runnable_component_list):
        self.runnable_component_list = runnable_component_list

    def invoke(self ,input_data):
        
        for runnable in self.runnable_component_list: # all components will invoke one by one
            input_data = runnable.invoke(input_data)  # every next component's input is the output of last component

        return input_data
```

```python
template = NakliPromptTemplate(
    template= "write a poem about {topic} of {length} length",
    input_variables=["length","topic"]
)
```

```python
llm  = NakliLLM()
```

```python
parser = NakliStrOutputParser()
```

```python
chain = RunnableConnector([template , llm , parser])
```

```python
chain.invoke({'length' : 'long' , "topic" : "india"})
```
