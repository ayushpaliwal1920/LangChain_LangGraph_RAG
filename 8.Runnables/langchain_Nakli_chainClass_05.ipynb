```python
import random

# Dummy LLm class : not logical just a demo

class NakliLLM:

    def __init__(self):
        print("LLM created")

    def predict(self, prompt):

        response_list = [
            "Delhi is the capital of INDIA.",
            "Ipl is a cricket league.",
            "Off campus is not that good except for lead female cast."
        ]

        return {"response" : random.choice(response_list)}
```

```python
llm = NakliLLM()
```

```python
llm.predict("What is the capital of India ?")
```

```python
# Dummy Prompt template class : just for demo

class NakliPromptTemplate :
    def __init__(self , template , input_variables):
        self.template = template
        self.input_variables = input_variables

    def format(self , input_dict):
        return self.template.format(**input_dict)
```

```python
template = NakliPromptTemplate(
    template= "write a poem about {topic} of {length} length",
    input_variables=["length","topic"]
)
```

```python
prompt = template.format({"topic" : 'india',
                 "length" : 'short'})
```

```python
llm = NakliLLM()
```

```python
llm.predict(prompt)
```

```python
# Dummy chain :

class NakliLLMChain :

    def __init__(self , llm , prompt):
        self.llm = llm
        self.prompt = prompt

    def run(self , input_dict):
        final_prompt = self.prompt.format(input_dict)
        result = self.llm.predict(final_prompt)

        return result['response']
```

```python
template2 = NakliPromptTemplate(
    template= "write a poem about {topic} of {length} length",
    input_variables=["length","topic"]
)
```

```python
llm2 = NakliLLM()
```

```python
chain = NakliLLMChain(llm2 , template2)
```

```python
chain.run({'length' : 'short', 'topic'  : 'india'})
```
