# Custom Tools  : using langchain 

```python
from langchain_core.tools import tool
```

```python
# Step 1 : Create a function 

def multiply(a , b):
    """ Multiply two integers and return the result. """  # Doc String highly recommended 

    return a*b
```

```python
# Step 2 : add type hints 

def multiply(a: int , b: int ) -> int :  # type hints added for input parameters and return type
    """ Multiply two integers and return the result. """
    return a*b
```

```python
# Step 3 : add tool decorator 

@tool
def multiply(a: int , b: int ) -> int :
    """ Multiply two integers and return the result. """
    return a*b
```

```python
result = multiply.invoke({"a":3 , "b":4})
```

```python
print(result)  # Output: 12
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

```python
print(multiply.args_schema.model_json_schema())
```

# Method 2 : Using Structured Tool 

```python
from langchain_core.tools import StructuredTool 
from pydantic import BaseModel , Field
```

```python
class MultiplyInput(BaseModel):
    a: int = Field(required = True , description = "The first number to add")
    b: int = Field(required = True , description = "The second number to add")
```

```python
def multiply_func(a : int , b: int) -> int :
    """ Multiply two integers and return the result. """
    return a*b

```

```python
multiply_tool = StructuredTool(
    func = multiply_func,
    name = "multiply",
    description = "Multiply two integers and return the result.",
    args_schema = MultiplyInput
)
```

```python
result = multiply_tool.invoke({"a": 5, "b": 6})

print(result)  # Output: 30
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args_schema.model_json_schema())
```

# Method 3 : using BaseTool class

```python
from langchain_core.tools import BaseTool
from typing import Type
```

```python
# arg schema using Pydantic

class MultiplyInput(BaseModel):
    a: int = Field(required = True , description = "The first number to add")
    b: int = Field(required = True , description = "The second number to add")
```

```python
class MultiplyTool(BaseTool):
    name: str = "multiply"
    description: str = "Multiply two integers and return the result."
    args_schema: Type[BaseModel] = MultiplyInput

    def _run(self , a:int , b:int) -> int:
        return a*b
```

```python
multiply_tool = MultiplyTool()
```

```python
result = multiply_tool.invoke({"a": 5, "b": 6})
print(result)  # Output: 30
print(multiply_tool.name)  
```

# ToolKit : 

```python
from langchain_core.tools import tool 

# custom tool 

@tool
def add(a: int , b: int ) -> int :
    """Add two numbers"""
    return a+b

@tool 
def Multiply(a: int , b: int ) -> int :
    """Multiply two numbers"""
    return a*b
```

```python
class MathToolkit:
    def get_tools(self):
        return [add,multiply]
```

```python
toolkit = MathToolkit()

tools = toolkit.get_tools()

for tool in tools:
    print(tool.name , "=>" , tool.description)
```
